# Agentic AI Health Analysis System - Demo

This notebook demonstrates the capabilities of the Agentic AI Health Assistant. It simulates a conversation with a user, tracks the internal state, and generates a final health report.

```python
import os
import json
from agent_logic import HealthAgent
from dotenv import load_dotenv

# Load API Key
load_dotenv()
if not os.getenv("GOOGLE_API_KEY"):
    print("WARNING: GOOGLE_API_KEY not found. Please set it in .env or environment.")
```

## Initialize Agent

```python
agent = HealthAgent(session_id="demo_user_01")
print("Agent initialized.")
```

## Simulation Loop

We will simulate a user reporting symptoms of Diabetes (thirst, frequent urination).

```python
def run_turn(user_input):
    print(f"\nUser: {user_input}")
    response = agent.step(user_input)
    print(f"Agent ({response['type']}):")
    print(json.dumps(response, indent=2))
    return response

# Turn 1: Initial Symptoms
resp1 = run_turn("I've been feeling very thirsty lately and I have to pee a lot, especially at night.")

# Turn 2: Answer Follow-up (likely about fatigue or weight)
if resp1['type'] == 'followup':
    # Simulate a generic answer based on common diabetes symptoms
    resp2 = run_turn("Yes, I feel tired all the time and I think I lost some weight without trying.")

# Turn 3: Answer Follow-up (likely about family history or vision)
if 'resp2' in locals() and resp2['type'] == 'followup':
    resp3 = run_turn("My father has type 2 diabetes. No blurry vision though.")

# Turn 4: Answer Follow-up (lifestyle)
if 'resp3' in locals() and resp3['type'] == 'followup':
    resp4 = run_turn("I don't exercise much and I eat a lot of sweets.")
```

## Check Internal State

```python
print("Current Risk Scores:")
print(json.dumps(agent.state.risk_scores, indent=2))

print("\nCollected Symptoms:")
print(agent.state.symptoms)
```
